```python
from pathlib import Path
import py_compile

base = Path("/content/AI")

# سازگاری ChatInterface با Gradio 6
for name in ["app.py", "chat.py"]:
    p = base / name
        s = p.read_text(encoding="utf-8")
            s = s.replace('type="messages",', "")
                s = s.replace(', type="messages"', "")
                    p.write_text(s, encoding="utf-8")

                    # اصلاح فعال‌سازی VAE tiling
                    p = base / "image.py"
                    s = p.read_text(encoding="utf-8")
                    s = s.replace("pipe.enable_vae_tiling()", "pipe.vae.enable_tiling()")
                    p.write_text(s, encoding="utf-8")

                    # بررسی نحوی تمام فایل‌ها
                    for name in ["app.py", "chat.py", "image.py"]:
                        py_compile.compile(str(base / name), doraise=True)
                            print(name, "OK")

                            print("اصلاحات اولیه و بررسی نحوی با موفقیت انجام شد.")
```
